## Now calculate it on the T points

```python
U_T = (U.U.shift(x=1) + U.U)*0.5

V_T = (V.V.shift(y=1) + V.V)*0.5
```

```python
UV = xr.Dataset(
    data_vars={
        'U': (('time','lat','lon'), U_T.data.compute()),
        'V': (('time','lat', 'lon'), V_T.data.compute()),
    },
    coords={
        'time':ssh.time_counter.data,
        'lat':ssh.nav_lat.isel(x=0).data,
        'lon':ssh.nav_lon.isel(y=0).data,
    }
)
UV
```

```python
UV['UV'] = (UV.U**2+UV.V**2)**0.5
UV.UV.isel(time=0).sel(lon=slice(-65,-40),lat=slice(-5,10)).plot()
```
