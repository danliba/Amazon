```python
## dummy example for stretching 
import numpy as np
import xarray as xr
```

```python
##dummy data

z = -10 #m particle depth
w_raw = 2e-6 #NEMO native W 
eta = 1 #SSH
H0 = 100 #bathymetry
u = 0.1 #uvel
v = 0.2 #vvel
detadlon = 0.01 #dssh/dlon --- m/deg
detadlat = 0.02 #dssh/lat --- m/deg
detadt_time = 0.2#m/s -- dssh/dt
```

```python
H = H0 + eta
detadt = detadt_time + u*detadlon + v*detadlat
stretch = detadt * (z+H0)/H

## corrected omega
omega = w_raw - stretch
omega
```
