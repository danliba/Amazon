## When to release particles

```python
import pandas as pd
import numpy as np

start = pd.Timestamp('1993-01-01')
end   = pd.Timestamp('2014-12-31')

# All 5-day releases
releases = pd.date_range(start, end, freq='5D')  # inclusive of start
```

### Periods/jobs of tracking for 2 years of TS and 10K particles
* First: 1993-01-06 to 1994-12-27
* Second: 1995-01-01 to 1996-12-31
* Third: 1997-01-05 to 1998-12-31
* Forth: 1999-01-05 to 2000-12-30

```python
import numpy as np

vals = np.array([d.strftime('%Y-%m-%d') for d in releases[350:150+450]])
table = vals.reshape(50, 5)  # rows, cols

for row in table:
    print('   '.join(row))
```
